### 🧠 **Bayes' Theorem: Sample Problem & Detailed Solution**

Let's work through a classic problem to solidify your understanding of Bayes' Theorem.

---

#### 💊 **Problem Scenario: Drug Testing**

A new drug test has been developed. Here are the statistics:

1.  **Prevalence:** 0.5% (0.005) of the population uses the drug.
2.  **Sensitivity (True Positive Rate):** The test correctly identifies drug users 99% of the time. ($P(\text{Positive Test} | \text{User}) = 0.99$)
3.  **Specificity (True Negative Rate):** The test correctly identifies non-users 98% of the time. ($P(\text{Negative Test} | \text{Non-User}) = 0.98$)

**Question:** If an individual tests positive, what is the probability that they actually use the drug?

---

#### 🎯 **Goal**

We want to find $P(\text{User} | \text{Positive Test})$.

---

#### 📝 **Step-by-Step Solution**

**1. Define Events & Probabilities:**

Let's define our events:
* $U$: The event that an individual **uses** the drug.
* $\neg U$: The event that an individual **does not use** the drug.
* $P$: The event that the test result is **positive**.
* $\neg P$: The event that the test result is **negative**.

Now, let's list the probabilities given in the problem:
* **Prior Probability of User:**
    $P(U) = 0.005$ (0.5% of the population uses the drug)

* **Prior Probability of Non-User:**
    $P(\neg U) = 1 - P(U) = 1 - 0.005 = 0.995$

* **Likelihood (True Positive Rate):**
    $P(P | U) = 0.99$ (Probability of a positive test given the person is a user)

* **Derive False Positive Rate (from Specificity):**
    Specificity is $P(\neg P | \neg U) = 0.98$.
    The **False Positive Rate** is the probability of a positive test given the person is *not* a user:
    $P(P | \neg U) = 1 - P(\neg P | \neg U) = 1 - 0.98 = 0.02$

---

**2. State Bayes' Theorem for Our Problem:**

Our goal is to find $P(U | P)$. Using the Bayes' Theorem formula:

$$
P(U | P) = \frac{P(P | U) \cdot P(U)}{P(P)}
$$

---

**3. Calculate $P(P)$ (Evidence Probability):**

The denominator, $P(P)$, is the total probability of getting a positive test result, regardless of whether the person uses the drug or not. We use the Law of Total Probability:

$$
P(P) = P(P | U) \cdot P(U) + P(P | \neg U) \cdot P(\neg U)
$$

Substitute the values we have:

$$
P(P) = (0.99 \cdot 0.005) + (0.02 \cdot 0.995)
$$

$$
P(P) = 0.00495 + 0.0199
$$

$$
P(P) = 0.02485
$$

*Explanation of $P(P)$:*
This value represents the overall probability that *anyone* picked randomly from the population will test positive. It's the sum of two scenarios:
1.  They are a user AND they test positive (true positive).
2.  They are a non-user AND they test positive (false positive).
Notice how the false positive path ($0.0199$) contributes significantly more to the overall positive test rate than the true positive path ($0.00495$) due to the high prevalence of non-users.

---

**4. Calculate $P(U | P)$ (Posterior Probability):**

Now, plug all the calculated values into the Bayes' Theorem formula:

$$
P(U | P) = \frac{P(P | U) \cdot P(U)}{P(P)}
$$

$$
P(U | P) = \frac{0.99 \cdot 0.005}{0.02485}
$$

$$
P(U | P) = \frac{0.00495}{0.02485}
$$

$$
P(U | P) \approx 0.199195
$$

---

#### 📊 **Conclusion & Interpretation**

If an individual tests positive, the probability that they actually use the drug is approximately **0.1992 (or about 19.92%)**.

**Why is this result surprising for many?**

* **Initial Intuition:** Many people, when first encountering this problem, might intuitively think that a 99% accurate test means a positive result indicates a near-certain chance of drug use.
* **The Power of Priors:** The key reason for this seemingly low posterior probability (19.92%) is the **low prior prevalence** of drug users in the population (0.5%).
    * Even though the test is quite accurate, the number of false positives (2% of the vast majority of non-users) far outweighs the number of true positives (99% of the tiny minority of users).
    * For every 1000 people:
        * 5 are users, 995 are non-users.
        * Of the 5 users, $0.99 \times 5 \approx 5$ will test positive (True Positives).
        * Of the 995 non-users, $0.02 \times 995 \approx 19.9$ will test positive (False Positives).
        * Total positive tests: $5 + 19.9 \approx 24.9$.
        * Probability of user given positive test: $\frac{\text{True Positives}}{\text{Total Positives}} = \frac{5}{24.9} \approx 0.20$.

This problem vividly demonstrates how Bayes' Theorem correctly balances the test's accuracy (likelihood) with the underlying base rate (prior probability) to give a realistic posterior probability. It's a crucial concept for understanding medical diagnoses, security screening, and many other real-world scenarios where rare events are involved.
