# 🧠 Naive Bayes Classifier — Mathematics & Derivation

---

## 🔹 Goal

Given:

* Features: $x = (x_1, x_2, ..., x_n)$
* Classes: $y \in \{y_1, y_2, ..., y_k\}$

We want to **find the most probable class** $\hat{y}$ for a given input $x$:

```math
\hat{y} = \arg\max_{y \in Y} P(y \mid x_1, x_2, ..., x_n)
```

---

## 🔹 Step 1: Apply Bayes' Theorem

Bayes' theorem:

```math
P(y \mid x) = \frac{P(x \mid y) \cdot P(y)}{P(x)}
```

In classification, we only care about which class has the highest probability, so we can ignore the constant denominator $P(x)$:

```math
\hat{y} = \arg\max_y P(x \mid y) \cdot P(y)
```

---

## 🔹 Step 2: The Naive Assumption

Assume **features are conditionally independent** given the class $y$:

```math
P(x \mid y) = P(x_1, x_2, ..., x_n \mid y) = \prod_{i=1}^{n} P(x_i \mid y)
```

Substitute this into the decision rule:

```math
\hat{y} = \arg\max_y P(y) \cdot \prod_{i=1}^{n} P(x_i \mid y)
```

This is the **Naive Bayes classification rule**.

---

## 🔹 Step 3: Use Log to Avoid Underflow

Since multiplying many small probabilities can cause underflow, we take the **log**:

```math
\hat{y} = \arg\max_y \left( \log P(y) + \sum_{i=1}^{n} \log P(x_i \mid y) \right)
```

---

## 🔸 Summary of Mathematical Steps

| Step                 | Formula                                                      |
| -------------------- | ------------------------------------------------------------ |
| **Bayes Rule**       | $P(y \mid x) = \frac{P(x \mid y) P(y)}{P(x)}$                |
| **Naive Assumption** | $P(x \mid y) = \prod_i P(x_i \mid y)$                        |
| **Final Rule**       | $\hat{y} = \arg\max_y P(y) \prod_i P(x_i \mid y)$            |
| **With log**         | $\hat{y} = \arg\max_y \log P(y) + \sum_i \log P(x_i \mid y)$ |

---

## 🔸 How Are the Probabilities Estimated?

* $P(y)$: Prior probability of class
  $\Rightarrow$ Estimated as:

  ```math
  P(y) = \frac{\text{Count of samples in class } y}{\text{Total samples}}
  ```

* $P(x_i \mid y)$: Likelihood of feature $x_i$ given class $y$
  Depends on the variant:

| Variant       | Feature Type | Estimation Technique               |
| ------------- | ------------ | ---------------------------------- |
| GaussianNB    | Continuous   | Estimate mean & variance per class |
| MultinomialNB | Count-based  | Word frequencies + smoothing       |
| BernoulliNB   | Binary       | Count of 0s and 1s per class       |

---

## 🔹 Example (Multinomial NB, like in text classification)

Let:

* $x_1$: count of word "AI"
* $x_2$: count of word "robot"
* $y \in \{\text{tech}, \text{sports}\}$

We compute:

```math
P(\text{tech}) \cdot P(\text{"AI"} \mid \text{tech})^{x_1} \cdot P(\text{"robot"} \mid \text{tech})^{x_2}
```

Then compare with class = "sports", and choose the class with the higher probability.

Let’s walk through the **full derivation and understanding of Naive Bayes** using a **concrete example**.

---

### 🧠 Problem: Email Spam Classification

Suppose we want to classify whether an email is **Spam** or **Not Spam (Ham)** based on two features:

| Email ID | "Buy" present? $x_1$ | "Win" present? $x_2$ | Class $y$ |
| -------- | -------------------- | -------------------- | --------- |
| 1        | Yes (1)              | Yes (1)              | Spam      |
| 2        | Yes (1)              | No  (0)              | Spam      |
| 3        | No  (0)              | Yes (1)              | Ham       |
| 4        | No  (0)              | No  (0)              | Ham       |
| 5        | Yes (1)              | Yes (1)              | Spam      |

We'll use **Bernoulli Naive Bayes** (binary features).

---

## 🔹 Step 1: Estimate Prior Probabilities

We have 5 emails:

* Spam = 3 emails → $P(\text{Spam}) = \frac{3}{5}$
* Ham = 2 emails → $P(\text{Ham}) = \frac{2}{5}$

---

## 🔹 Step 2: Estimate Likelihoods

For each feature and class, compute:

```math
P(x_i = 1 \mid y)
```

### For Class = Spam:

* Emails: 1, 2, 5

| Feature | Count where present | Total | Smoothed (Laplace)                  |
| ------- | ------------------- | ----- | ----------------------------------- |
| "Buy"   | 3                   | 3     | $\frac{3 + 1}{3 + 2} = \frac{4}{5}$ |
| "Win"   | 2                   | 3     | $\frac{2 + 1}{3 + 2} = \frac{3}{5}$ |

### For Class = Ham:

* Emails: 3, 4

| Feature | Count where present | Total | Smoothed                                  |
| ------- | ------------------- | ----- | ----------------------------------------- |
| "Buy"   | 0                   | 2     | $\frac{0 + 1}{2 + 2} = \frac{1}{4}$       |
| "Win"   | 1                   | 2     | $\frac{1 + 1}{2 + 2} = \frac{2}{4} = 0.5$ |

---

## 🔹 Step 3: Make Prediction

Let’s classify a **new email** with:

* "Buy" = Yes (1)
* "Win" = Yes (1)
  ⇒ $x = (1, 1)$

We compute:

### For Spam:

```math
P(\text{Spam}) \cdot P(\text{"Buy"} = 1 \mid \text{Spam}) \cdot P(\text{"Win"} = 1 \mid \text{Spam})
= \frac{3}{5} \cdot \frac{4}{5} \cdot \frac{3}{5}
= \frac{36}{125} = 0.288
```

### For Ham:

```math
P(\text{Ham}) \cdot P(\text{"Buy"} = 1 \mid \text{Ham}) \cdot P(\text{"Win"} = 1 \mid \text{Ham})
= \frac{2}{5} \cdot \frac{1}{4} \cdot \frac{1}{2}
= \frac{1}{20} = 0.05
```

---

## ✅ Final Prediction

Since:

```math
P(\text{Spam} \mid x) > P(\text{Ham} \mid x)
```

→ **Predicted class = Spam**

---

## 🔹 Full Generalized Formula (Bernoulli NB)

```math
\hat{y} = \arg\max_y \log P(y) + \sum_{i=1}^{n} x_i \log P(x_i = 1 \mid y) + (1 - x_i) \log P(x_i = 0 \mid y)
```

This formulation is used internally to handle both presence and absence of features.

---

## 🧠 Summary

| Step | Description                           |
| ---- | ------------------------------------- |
| 1️⃣  | Estimate $P(y)$: class prior          |
| 2️⃣  | Estimate $P(x_i \mid y)$: likelihoods |
| 3️⃣  | Plug into Naive Bayes formula         |
| 4️⃣  | Use log for numerical stability       |
| ✅    | Pick class with highest probability   |
