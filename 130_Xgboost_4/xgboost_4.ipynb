# 🚀 XGBoost Mathematics: Complete Guide
## Classification & Regression with Examples

---

## 📊 **Overview**
XGBoost (eXtreme Gradient Boosting) is an ensemble method that builds models sequentially, where each new model corrects errors from previous models using gradient boosting.

---

## 🎯 **Core Mathematical Foundation**

### **1. Objective Function**
The general objective function XGBoost minimizes:

$$\mathcal{L}(\phi) = \sum_{i=1}^{n} l(y_i, \hat{y}_i) + \sum_{k=1}^{K} \Omega(f_k)$$

Where:
- $l(y_i, \hat{y}_i)$ = Loss function (training loss)
- $\Omega(f_k)$ = Regularization term for tree $k$
- $K$ = Number of trees
- $n$ = Number of samples

### **2. Regularization Term**
$$\Omega(f) = \gamma T + \frac{1}{2}\lambda \sum_{j=1}^{T} w_j^2$$

Where:
- $\gamma$ = Minimum loss reduction required for split
- $T$ = Number of leaves
- $\lambda$ = L2 regularization parameter
- $w_j$ = Weight of leaf $j$

---

## 🔄 **Gradient Boosting Process**

### **Step-by-Step Algorithm**

**Initial Prediction:**
$$\hat{y}_i^{(0)} = \text{const}$$

**For iteration $t = 1, 2, ..., T$:**

1. **Calculate Gradients:**
   $$g_i = \frac{\partial l(y_i, \hat{y}_i^{(t-1)})}{\partial \hat{y}_i^{(t-1)}}$$

2. **Calculate Hessians:**
   $$h_i = \frac{\partial^2 l(y_i, \hat{y}_i^{(t-1)})}{\partial (\hat{y}_i^{(t-1)})^2}$$

3. **Build Tree $f_t$:**
   - Find optimal splits using gain formula
   - Assign leaf weights

4. **Update Predictions:**
   $$\hat{y}_i^{(t)} = \hat{y}_i^{(t-1)} + \eta \cdot f_t(x_i)$$

Where $\eta$ is the learning rate.

---

## 📈 **REGRESSION MATHEMATICS**

### **Loss Function: Mean Squared Error**
$$l(y_i, \hat{y}_i) = \frac{1}{2}(y_i - \hat{y}_i)^2$$

### **Gradients and Hessians**
- **First Derivative (Gradient):**
  $$g_i = \frac{\partial l}{\partial \hat{y}_i} = \hat{y}_i - y_i$$

- **Second Derivative (Hessian):**
  $$h_i = \frac{\partial^2 l}{\partial \hat{y}_i^2} = 1$$

### **📝 Detailed Example: Regression**

**Dataset:**
| Sample | Age | Income | House Price |
|--------|-----|--------|-------------|
| 1      | 25  | 50k    | 200k        |
| 2      | 35  | 80k    | 350k        |
| 3      | 45  | 120k   | 500k        |

**Step 1: Initial Prediction**
$$\hat{y}^{(0)} = \text{mean}(y) = \frac{200 + 350 + 500}{3} = 350$$

**Step 2: Calculate Residuals (Gradients)**
- $g_1 = 350 - 200 = 150$
- $g_2 = 350 - 350 = 0$
- $g_3 = 350 - 500 = -150$

**Step 3: Build First Tree**
Since $h_i = 1$ for all samples, we find the optimal leaf weight:

$$w^* = -\frac{\sum_{i \in I_j} g_i}{\sum_{i \in I_j} h_i + \lambda}$$

For a leaf containing samples 1 and 3:
$$w^* = -\frac{150 + (-150)}{1 + 1 + \lambda} = \frac{0}{2 + \lambda} = 0$$

**Step 4: Update Predictions**
$$\hat{y}^{(1)} = \hat{y}^{(0)} + \eta \cdot f_1(x)$$

---

## 🎲 **CLASSIFICATION MATHEMATICS**

### **Binary Classification: Logistic Loss**
$$l(y_i, \hat{y}_i) = y_i \log(1 + e^{-\hat{y}_i}) + (1-y_i) \log(1 + e^{\hat{y}_i})$$

### **Probability Transformation**
$$p_i = \frac{1}{1 + e^{-\hat{y}_i}}$$

### **Gradients and Hessians**
- **First Derivative:**
  $$g_i = p_i - y_i$$

- **Second Derivative:**
  $$h_i = p_i(1 - p_i)$$

### **📝 Detailed Example: Binary Classification**

**Dataset:**
| Sample | Feature 1 | Feature 2 | Label |
|--------|-----------|-----------|-------|
| 1      | 2.5       | 3.1       | 1     |
| 2      | 1.8       | 2.2       | 0     |
| 3      | 3.2       | 4.5       | 1     |

**Step 1: Initial Prediction**
$$\hat{y}^{(0)} = \log\left(\frac{\text{positive}}{\text{negative}}\right) = \log\left(\frac{2}{1}\right) = \log(2) \approx 0.693$$

**Step 2: Calculate Probabilities**
$$p_i^{(0)} = \frac{1}{1 + e^{-0.693}} = \frac{1}{1 + 0.5} = 0.667$$

**Step 3: Calculate Gradients**
- $g_1 = 0.667 - 1 = -0.333$
- $g_2 = 0.667 - 0 = 0.667$
- $g_3 = 0.667 - 1 = -0.333$

**Step 4: Calculate Hessians**
- $h_i = 0.667(1 - 0.667) = 0.222$ for all samples

**Step 5: Build Tree and Update**
Optimal leaf weight:
$$w^* = -\frac{\sum g_i}{\sum h_i + \lambda}$$

---

## 🌳 **Tree Construction Mathematics**

### **Splitting Criterion: Gain Formula**
$$\text{Gain} = \frac{1}{2}\left[\frac{(\sum_{i \in I_L} g_i)^2}{\sum_{i \in I_L} h_i + \lambda} + \frac{(\sum_{i \in I_R} g_i)^2}{\sum_{i \in I_R} h_i + \lambda} - \frac{(\sum_{i \in I} g_i)^2}{\sum_{i \in I} h_i + \lambda}\right] - \gamma$$

Where:
- $I_L$ = Left child samples
- $I_R$ = Right child samples
- $I$ = Parent node samples

### **Optimal Leaf Weight**
$$w_j^* = -\frac{\sum_{i \in I_j} g_i}{\sum_{i \in I_j} h_i + \lambda}$$

### **📝 Splitting Example**

**Current Node:** 3 samples with gradients $g = [150, 0, -150]$ and hessians $h = [1, 1, 1]$

**Potential Split:** Age < 30
- **Left:** Sample 1, $g_L = 150$, $h_L = 1$
- **Right:** Samples 2,3, $g_R = 0 + (-150) = -150$, $h_R = 2$

**Calculate Gain:**
$$\text{Gain} = \frac{1}{2}\left[\frac{150^2}{1 + \lambda} + \frac{(-150)^2}{2 + \lambda} - \frac{0^2}{3 + \lambda}\right] - \gamma$$

Assuming $\lambda = 0$:
$$\text{Gain} = \frac{1}{2}\left[\frac{22500}{1} + \frac{22500}{2} - 0\right] - \gamma = \frac{1}{2}[22500 + 11250] - \gamma = 16875 - \gamma$$

---

## 📊 **Multi-class Classification**

### **Softmax Loss Function**
$$l(y, \hat{y}) = -\sum_{k=1}^{K} y_k \log(p_k)$$

Where:
$$p_k = \frac{e^{\hat{y}_k}}{\sum_{j=1}^{K} e^{\hat{y}_j}}$$

### **Gradients for Class $k$**
$$g_{i,k} = p_{i,k} - y_{i,k}$$

### **Hessians for Class $k$**
$$h_{i,k} = p_{i,k}(1 - p_{i,k})$$

---

## ⚙️ **Hyperparameter Impact**

### **Learning Rate ($\eta$)**
- Controls step size in gradient descent
- Lower values → More trees needed, better generalization
- Higher values → Faster training, risk of overfitting

### **Regularization Parameters**
- **$\lambda$ (L2):** Penalizes large leaf weights
- **$\gamma$ (Min split loss):** Prevents overfitting by requiring minimum gain
- **Max depth:** Limits tree complexity

### **📝 Mathematical Impact Example**

**Without Regularization ($\lambda = 0$):**
$$w^* = -\frac{\sum g_i}{\sum h_i}$$

**With Regularization ($\lambda = 1$):**
$$w^* = -\frac{\sum g_i}{\sum h_i + 1}$$

If $\sum g_i = 10$ and $\sum h_i = 5$:
- Without: $w^* = -\frac{10}{5} = -2$
- With: $w^* = -\frac{10}{6} = -1.67$

Regularization reduces the magnitude of updates, preventing overfitting.

## 🎯 **Key Takeaways**

1. **XGBoost uses second-order gradients (Hessians)** for more accurate approximation
2. **Regularization is built into the objective function** to prevent overfitting
3. **Tree construction uses a mathematically principled gain formula**
4. **Different loss functions adapt the algorithm** to various tasks
5. **Hyperparameters directly influence** the mathematical optimization process

# 🚀 XGBoost: Complete Mathematical Guide
## All the Math You Need to Understand XGBoost

---

## 📊 **CORE CONCEPT: What is XGBoost?**

XGBoost builds an ensemble of decision trees sequentially, where each new tree corrects errors from previous trees. The final prediction is:

$$\hat{y}_i = \sum_{k=1}^{K} f_k(x_i)$$

Where:
- $f_k$ = $k$-th tree
- $K$ = total number of trees
- Each tree $f_k$ maps input $x_i$ to a leaf weight

---

## 🎯 **OBJECTIVE FUNCTION**

### **Complete Objective Function**
$$\mathcal{L} = \sum_{i=1}^{n} l(y_i, \hat{y}_i) + \sum_{k=1}^{K} \Omega(f_k)$$

**Components:**
- **Training Loss:** $\sum_{i=1}^{n} l(y_i, \hat{y}_i)$ - How well we fit the data
- **Regularization:** $\sum_{k=1}^{K} \Omega(f_k)$ - Prevents overfitting

### **Regularization Term**
$$\Omega(f) = \gamma T + \frac{1}{2}\lambda \sum_{j=1}^{T} w_j^2$$

Where:
- $T$ = Number of leaves in the tree
- $w_j$ = Weight of leaf $j$
- $\gamma$ = Penalty for each leaf (complexity cost)
- $\lambda$ = L2 regularization parameter

**Intuition:** 
- $\gamma T$ penalizes complex trees (more leaves)
- $\frac{1}{2}\lambda \sum w_j^2$ penalizes large leaf weights

---

## 🔄 **GRADIENT BOOSTING ALGORITHM**

### **Step-by-Step Process**

**1. Initialize:**
$$F_0(x) = \arg\min_{\gamma} \sum_{i=1}^{n} l(y_i, \gamma)$$

**2. For each iteration $t = 1, 2, ..., T$:**

**a) Calculate gradients:**
$$g_i = \frac{\partial l(y_i, F_{t-1}(x_i))}{\partial F_{t-1}(x_i)}$$

**b) Calculate Hessians:**
$$h_i = \frac{\partial^2 l(y_i, F_{t-1}(x_i))}{\partial F_{t-1}(x_i)^2}$$

**c) Build tree $f_t$ to minimize:**
$$\sum_{i=1}^{n} [g_i f_t(x_i) + \frac{1}{2}h_i f_t^2(x_i)] + \Omega(f_t)$$

**d) Update model:**
$$F_t(x) = F_{t-1}(x) + \eta f_t(x)$$

Where $\eta$ is the learning rate.

---

## 🌳 **TREE BUILDING MATHEMATICS**

### **Key Insight: Second-Order Approximation**

Instead of just using gradients (first derivatives), XGBoost uses **both gradients and Hessians** (second derivatives). This is like using **Newton's method** instead of gradient descent.

**Taylor Expansion:**
$$l(y_i, F_{t-1}(x_i) + f_t(x_i)) \approx l(y_i, F_{t-1}(x_i)) + g_i f_t(x_i) + \frac{1}{2}h_i f_t^2(x_i)$$

### **Reformulating for Tree Structure**

For a tree with $T$ leaves, let $I_j = \{i | q(x_i) = j\}$ be the set of samples in leaf $j$:

$$\mathcal{L}^{(t)} = \sum_{j=1}^{T} \left[(\sum_{i \in I_j} g_i) w_j + \frac{1}{2}(\sum_{i \in I_j} h_i + \lambda) w_j^2\right] + \gamma T$$

**Simplifying notation:**
- $G_j = \sum_{i \in I_j} g_i$ (sum of gradients in leaf $j$)
- $H_j = \sum_{i \in I_j} h_i$ (sum of Hessians in leaf $j$)

$$\mathcal{L}^{(t)} = \sum_{j=1}^{T} \left[G_j w_j + \frac{1}{2}(H_j + \lambda) w_j^2\right] + \gamma T$$

---

## 🔧 **OPTIMAL LEAF WEIGHTS**

### **Finding the Best Leaf Weight**

For each leaf $j$, we minimize:
$$\mathcal{L}_j = G_j w_j + \frac{1}{2}(H_j + \lambda) w_j^2$$

Taking derivative and setting to zero:
$$\frac{\partial \mathcal{L}_j}{\partial w_j} = G_j + (H_j + \lambda) w_j = 0$$

**Optimal leaf weight:**
$$w_j^* = -\frac{G_j}{H_j + \lambda}$$

**Intuition:** 
- If $G_j > 0$ (predictions too high), weight is negative
- If $G_j < 0$ (predictions too low), weight is positive
- $\lambda$ prevents weights from becoming too large

---

## 📏 **SPLITTING CRITERIA**

### **The Gain Formula**

When we split a node into left and right children, the **gain** is:

$$Gain = \frac{1}{2}\left[\frac{G_L^2}{H_L + \lambda} + \frac{G_R^2}{H_R + \lambda} - \frac{G^2}{H + \lambda}\right] - \gamma$$

Where:
- $G_L, H_L$ = gradient and Hessian sums for left child
- $G_R, H_R$ = gradient and Hessian sums for right child  
- $G, H$ = gradient and Hessian sums for parent node
- $\gamma$ = minimum gain required to make a split

### **Why This Formula Works**

**Before split:** Optimal objective = $-\frac{1}{2} \cdot \frac{G^2}{H + \lambda}$

**After split:** Optimal objective = $-\frac{1}{2}\left[\frac{G_L^2}{H_L + \lambda} + \frac{G_R^2}{H_R + \lambda}\right]$

**Gain = Reduction in objective** = (After split) - (Before split)

### **📝 Detailed Splitting Example**

**Current Node:**
- 4 samples with gradients: $g = [10, -5, 8, -3]$
- All Hessians = 1: $h = [1, 1, 1, 1]$
- $G = 10 + (-5) + 8 + (-3) = 10$
- $H = 1 + 1 + 1 + 1 = 4$

**Potential Split:** Feature A < 2.5
- **Left child:** Samples 1, 3 → $G_L = 10 + 8 = 18$, $H_L = 2$
- **Right child:** Samples 2, 4 → $G_R = -5 + (-3) = -8$, $H_R = 2$

**Calculate Gain (assume $\lambda = 1$, $\gamma = 0$):**
$$Gain = \frac{1}{2}\left[\frac{18^2}{2 + 1} + \frac{(-8)^2}{2 + 1} - \frac{10^2}{4 + 1}\right]$$
$$= \frac{1}{2}\left[\frac{324}{3} + \frac{64}{3} - \frac{100}{5}\right]$$
$$= \frac{1}{2}[108 + 21.33 - 20] = \frac{1}{2} \times 109.33 = 54.67$$

**Since Gain > 0, this split improves the model!**

## 📊 **REGRESSION MATHEMATICS**

### **Loss Function: Squared Error**
$$l(y, \hat{y}) = \frac{1}{2}(y - \hat{y})^2$$

**Gradients and Hessians:**
- $g_i = \frac{\partial l}{\partial \hat{y}_i} = \hat{y}_i - y_i$ (residual)
- $h_i = \frac{\partial^2 l}{\partial \hat{y}_i^2} = 1$ (constant)

### **📝 Complete Regression Example**

**Dataset:** Predicting house prices
| Sample | Features | Actual Price | Current Pred | Residual |
|--------|----------|--------------|--------------|----------|
| 1      | [2,1]    | 300k         | 280k         | -20k     |
| 2      | [3,2]    | 450k         | 420k         | -30k     |
| 3      | [4,2]    | 600k         | 650k         | +50k     |
| 4      | [1,1]    | 200k         | 180k         | -20k     |

**Gradients:**
- $g_1 = 280 - 300 = -20$
- $g_2 = 420 - 450 = -30$  
- $g_3 = 650 - 600 = +50$
- $g_4 = 180 - 200 = -20$

**Hessians:** $h_i = 1$ for all samples

**Tree Building:**
Consider split: Bedrooms < 2.5
- **Left:** Samples 1, 4 → $G_L = -20 + (-20) = -40$, $H_L = 2$
- **Right:** Samples 2, 3 → $G_R = -30 + 50 = 20$, $H_R = 2$

**Gain calculation (λ = 1):**
$$Gain = \frac{1}{2}\left[\frac{(-40)^2}{3} + \frac{20^2}{3} - \frac{(-20)^2}{5}\right] = \frac{1}{2}[533.33 + 133.33 - 80] = 293.33$$

**Optimal leaf weights:**
- Left: $w_L^* = -\frac{-40}{2+1} = \frac{40}{3} = 13.33$
- Right: $w_R^* = -\frac{20}{2+1} = -\frac{20}{3} = -6.67$

**Interpretation:** 
- Houses with ≤2 bedrooms: increase prediction by 13.33k
- Houses with >2 bedrooms: decrease prediction by 6.67k

## 🎯 **CLASSIFICATION MATHEMATICS**

### **Binary Classification: Logistic Loss**
$$l(y, \hat{y}) = y \log(1 + e^{-\hat{y}}) + (1-y) \log(1 + e^{\hat{y}})$$

**Probability transformation:**
$$p = \frac{1}{1 + e^{-\hat{y}}}$$

**Gradients and Hessians:**
- $g_i = p_i - y_i$ (prediction error)
- $h_i = p_i(1 - p_i)$ (variance of prediction)

### **📝 Complete Classification Example**

**Dataset:** Email spam detection
| Sample | Features | Spam | Current F(x) | Probability | Gradient | Hessian |
|--------|----------|------|--------------|-------------|----------|---------|
| 1      | [5,2]    | 0    | 0.5          | 0.62        | 0.62     | 0.24    |
| 2      | [10,8]   | 1    | 0.5          | 0.62        | -0.38    | 0.24    |
| 3      | [2,1]    | 0    | 0.5          | 0.62        | 0.62     | 0.24    |
| 4      | [12,10]  | 1    | 0.5          | 0.62        | -0.38    | 0.24    |

**Tree Building:**
Split: Word count < 7
- **Left:** Samples 1, 3 → $G_L = 0.62 + 0.62 = 1.24$, $H_L = 0.48$
- **Right:** Samples 2, 4 → $G_R = -0.38 + (-0.38) = -0.76$, $H_R = 0.48$

**Gain calculation (λ = 0.1):**
$$Gain = \frac{1}{2}\left[\frac{1.24^2}{0.48 + 0.1} + \frac{(-0.76)^2}{0.48 + 0.1} - \frac{0.48^2}{0.96 + 0.1}\right]$$
$$= \frac{1}{2}[2.65 + 0.99 - 0.22] = 1.71$$

**Optimal leaf weights:**
- Left: $w_L^* = -\frac{1.24}{0.58} = -2.14$
- Right: $w_R^* = -\frac{-0.76}{0.58} = 1.31$

**Interpretation:**
- Low word count emails: decrease spam probability
- High word count emails: increase spam probability

## 🔄 **MULTICLASS CLASSIFICATION**

### **Softmax Loss**
For $K$ classes, XGBoost trains $K$ separate models:

$$p_k = \frac{e^{F_k(x)}}{\sum_{j=1}^{K} e^{F_j(x)}}$$

**Loss function:**
$$l(y, \hat{y}) = -\sum_{k=1}^{K} y_k \log(p_k)$$

**Gradients and Hessians:**
- $g_{i,k} = p_{i,k} - y_{i,k}$
- $h_{i,k} = p_{i,k}(1 - p_{i,k})$

## 📈 **ADVANCED SPLITTING METHODS**

### **1. Exact Greedy Algorithm**
For each feature and each possible split point:
1. Calculate gain using the gain formula
2. Choose split with maximum gain
3. **Time complexity:** $O(n \times d \times \log n)$

### **2. Approximate Algorithm**
For large datasets, use **quantile-based splitting**:

**Weighted quantile sketch:**
- Create candidate split points based on percentiles
- Weight each data point by its Hessian value
- **Why Hessian weighting?** Points with higher Hessian contribute more to the loss

**Quantile definition:**
$$r_k(z) = \frac{1}{\sum_{(x,h) \in D_k} h} \sum_{(x,h) \in D_k, x < z} h$$

Where $D_k$ is the dataset for feature $k$.

### **3. Sparsity-Aware Algorithm**
For missing values, XGBoost learns **default directions**:

**For each split, try both:**
1. Send missing values to left child
2. Send missing values to right child

**Choose the direction that gives higher gain.**

## 🛠️ **REGULARIZATION TECHNIQUES**

### **1. Tree Structure Regularization**
- **Max depth:** Limits tree depth
- **Min child weight:** Requires $\sum_{i \in I_j} h_i \geq \text{min_child_weight}$
- **Gamma (γ):** Minimum gain required for split

### **2. Stochastic Methods**
- **Subsample:** Use random subset of training data
- **Column sampling:** Use random subset of features
- **Column sampling by level:** Different features at each tree level

### **3. Shrinkage**
- **Learning rate (η):** Scale each tree's contribution
- **Early stopping:** Stop when validation error increases

## 📊 **LOSS FUNCTIONS COMPARISON**

### **Regression Loss Functions**

**1. Squared Error (default):**
- $l(y, \hat{y}) = \frac{1}{2}(y - \hat{y})^2$
- $g_i = \hat{y}_i - y_i$, $h_i = 1$

**2. Absolute Error:**
- $l(y, \hat{y}) = |y - \hat{y}|$
- $g_i = \text{sign}(\hat{y}_i - y_i)$, $h_i = 0$

**3. Huber Loss:**
- Combines squared and absolute error
- Robust to outliers

### **Classification Loss Functions**

**1. Logistic Loss (default for binary):**
- $l(y, \hat{y}) = \log(1 + e^{-y\hat{y}})$
- $g_i = -\frac{y_i}{1 + e^{y_i\hat{y}_i}}$

**2. Hinge Loss:**
- $l(y, \hat{y}) = \max(0, 1 - y\hat{y})$
- Used for SVM-like classification

---

## 🎯 **HYPERPARAMETER MATHEMATICAL IMPACT**

### **Learning Rate (η)**
$$F_t(x) = F_{t-1}(x) + \eta \cdot f_t(x)$$

**Effect on convergence:**
- Small η: Slow convergence, better generalization
- Large η: Fast convergence, risk of overfitting

### **Regularization Parameters**
**Lambda (λ):**
$$w_j^* = -\frac{G_j}{H_j + \lambda}$$

**Effect:** Higher λ → Smaller leaf weights → Less overfitting

**Gamma (γ):**
$$\text{Split if } Gain > \gamma$$

**Effect:** Higher γ → Fewer splits → Simpler trees

### **Tree Structure Parameters**
**Max depth:** Controls tree complexity
**Min child weight:** Prevents overfitting on small samples

---

## 🔍 **PRACTICAL CONSIDERATIONS**

### **1. Handling Imbalanced Data**
- **Scale pos weight:** Multiply positive class loss by factor
- **Focal loss:** Focus on hard examples

### **2. Feature Importance**
**Gain-based importance:**
$$\text{Importance}_j = \sum_{t=1}^{T} \sum_{s \in S_t} \mathbb{I}(f_s = j) \cdot Gain_s$$

Where $S_t$ is the set of splits in tree $t$.

### **3. Cross-Validation**
Use k-fold CV to:
- Select optimal number of rounds
- Tune hyperparameters
- Estimate generalization error

## 📚 **ALGORITHM SUMMARY**

### **XGBoost Training Algorithm**

```
1. Initialize F_0(x) = argmin_γ Σ l(y_i, γ)

2. For t = 1 to T:
   a. Compute gradients: g_i = ∂l(y_i, F_{t-1}(x_i))/∂F_{t-1}(x_i)
   b. Compute Hessians: h_i = ∂²l(y_i, F_{t-1}(x_i))/∂F_{t-1}(x_i)²
   
   c. Build tree f_t:
      - For each leaf j, compute optimal weight: w_j* = -G_j/(H_j + λ)
      - For each split, compute gain and choose best split
      - Stop splitting when gain < γ
   
   d. Update: F_t(x) = F_{t-1}(x) + η·f_t(x)

3. Return F_T(x)
```

### **Prediction Algorithm**

```
1. For each tree k = 1 to K:
   a. Traverse tree to find leaf
   b. Add leaf weight to prediction

2. For regression: return sum
3. For classification: return sigmoid(sum) or softmax(sum)
```

| **Regression Gradient** | $g_i = \hat{y}_i - y_i$ | Residual |
| **Classification Gradient** | $g_i = p_i - y_i$ | Prediction error |
| **Classification Hessian** | $h_i = p_i(1-p_i)$ | Prediction variance |
