### 📚 **Bayes' Theorem**

Bayes' Theorem is a fundamental concept in probability theory and statistics that describes how to update the probability of a hypothesis ($H$) when new evidence ($E$) is introduced. It's particularly powerful in machine learning for tasks like classification and inference.

---

#### 💡 **Core Idea**

It allows us to calculate `posterior probability` ($P(H|E)$) based on `prior probability` ($P(H)$), `likelihood` ($P(E|H)$), and `evidence probability` ($P(E)$).

---

#### 📐 **The Formula**

The theorem is mathematically expressed as:

$$
P(H|E) = \frac{P(E|H) \cdot P(H)}{P(E)}
$$

Where:
* $P(H|E)$: **Posterior Probability**
    * The probability of hypothesis $H$ being true, given the evidence $E$. This is what we want to find.
* $P(E|H)$: **Likelihood**
    * The probability of observing the evidence $E$, given that the hypothesis $H$ is true. This tells us how likely the evidence is under the hypothesis.
* $P(H)$: **Prior Probability**
    * The initial probability of hypothesis $H$ being true before any evidence is observed. This represents our prior belief.
* $P(E)$: **Evidence Probability (Marginal Likelihood)**
    * The total probability of observing the evidence $E$, regardless of the hypothesis. It acts as a normalizing constant to ensure the posterior probability is a valid probability (sums to 1).

---

#### 🌟 **Expanding $P(E)$**

The denominator $P(E)$ can often be hard to calculate directly. We can expand it using the law of total probability:

$$
P(E) = \sum_{i} P(E|H_i) \cdot P(H_i)
$$

For a binary hypothesis (e.g., $H$ and not $H$, denoted as $\neg H$ or $H^c$):

$$
P(E) = P(E|H) \cdot P(H) + P(E|\neg H) \cdot P(\neg H)
$$

Substituting this back into the main formula gives:

$$
P(H|E) = \frac{P(E|H) \cdot P(H)}{P(E|H) \cdot P(H) + P(E|\neg H) \cdot P(\neg H)}
$$

---

#### 🧠 **Intuition & Interpretation**

* **Updating Beliefs:** Bayes' Theorem provides a principled way to update our beliefs (prior) about a hypothesis in light of new data (evidence).
* **Prior vs. Posterior:** The prior $P(H)$ is our initial guess. The likelihood $P(E|H)$ quantifies how well the hypothesis explains the data. The posterior $P(H|E)$ is our updated, more informed belief.
* **"Inverse Probability":** It's sometimes called "inverse probability" because it allows us to infer the probability of a cause (hypothesis) given an effect (evidence).

---

#### 💡 **Applications in ML**

1.  **Naïve Bayes Classifiers:**
    * Widely used for text classification (spam detection, sentiment analysis).
    * It simplifies the calculation by assuming conditional independence between features, given the class.

2.  **Bayesian Inference:**
    * Used for estimating parameters of a model.
    * It provides a distribution over the parameters rather than a single point estimate, reflecting uncertainty.

3.  **Spam Filtering:**
    * Classic example: calculating the probability that an email is spam given certain words appear in it.

4.  **Medical Diagnosis:**
    * Calculating the probability of a disease given certain symptoms.

---

#### 📝 **Example Scenario**

Imagine a rare disease ($D$) that affects 1% of the population. A test for the disease ($T$) has:
* 95% true positive rate ($P(T|D) = 0.95$)
* 10% false positive rate ($P(T|\neg D) = 0.10$)

If a person tests positive ($T$), what is the probability they actually have the disease ($P(D|T)$)?

Given:
* $P(D) = 0.01$ (Prior probability of having the disease)
* $P(\neg D) = 1 - P(D) = 0.99$ (Prior probability of not having the disease)
* $P(T|D) = 0.95$ (Likelihood of positive test given disease)
* $P(T|\neg D) = 0.10$ (Likelihood of positive test given no disease)

Using Bayes' Theorem:

$$
P(D|T) = \frac{P(T|D) \cdot P(D)}{P(T|D) \cdot P(D) + P(T|\neg D) \cdot P(\neg D)}
$$

$$
P(D|T) = \frac{0.95 \cdot 0.01}{0.95 \cdot 0.01 + 0.10 \cdot 0.99}
$$

$$
P(D|T) = \frac{0.0095}{0.0095 + 0.099}
$$

$$
P(D|T) = \frac{0.0095}{0.1085} \approx 0.0875
$$

So, even with a positive test, there's only about an 8.75% chance of actually having the disease, due to its rarity and the test's false positive rate. This highlights the importance of incorporating prior probabilities.
